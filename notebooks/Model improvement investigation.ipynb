```python

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
df=pd.read_csv("Used_Car_Price_Prediction.csv")
display(df.head())
print(df.isnull().sum())
```

```python
df = df[df['sale_price'] > 0].copy()
```

```python
df=df.drop(columns= [
    'car_name',
    'times_viewed',
    'variant',
    'assured_buy',
    'registered_city',
    'registered_state',
    'is_hot',
    'rto',
    'source',
    'car_availability',
    'broker_quote',
    
    'car_rating',
    'ad_created_on',
    'fitness_certificate',
    'emi_starts_from',
    'booking_down_pymnt',
    'reserved',
    'warranty_avail'
])
```

```python
df['body_type']=df['body_type'].fillna(df['body_type'].mode()[0])
df['transmission']=df['transmission'].fillna(df['transmission'].mode()[0])
print(df.isnull().sum())
```

```python
print(df.isnull().sum())
print(df['original_price'].describe())
#df['original_price'] = df['original_price'].fillna(df['original_price'].median())
df=df.dropna()
print(df.isnull().sum())
```

```python
print(df.shape)
```

```python
print(
    df[['original_price', 'sale_price']]
    .corr()
)
print(df.isnull().sum())
```

```python
x=df.drop('sale_price',axis=1)
y=df['sale_price']
x=pd.get_dummies(x,columns=['fuel_type', 'city', 'body_type', 'transmission', 'make', 'model'],dtype=int)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
lr=LinearRegression()
print(x.isnull().sum())
lr.fit(x_train,y_train)
```

```python
y_pre=lr.predict(x_test)
j=pd.DataFrame({"act":y_test,"pre":y_pre})
print(j)
```

```python
from sklearn import metrics

mae = metrics.mean_absolute_error(y_test, y_pre)
mse = metrics.mean_squared_error(y_test, y_pre)
rmse = metrics.root_mean_squared_error(y_test, y_pre)
r2 = metrics.r2_score(y_test, y_pre)

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R2  :", r2)
```
