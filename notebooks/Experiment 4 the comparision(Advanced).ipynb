```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn import metrics
```

```python
df=pd.read_csv("Used_Car_Price_Prediction.csv")
display(df.head())
print(df.isnull().sum())
```

```python
columns = [
    'yr_mfr',
    'fuel_type',
    'kms_run',
    'sale_price',
    'city',
    'body_type',
    'transmission',
    'make',
    'model',
    'total_owners',
    'original_price'
]

df = df[columns].copy()
```

```python
df = df[df['sale_price'] > 0].copy()
```

```python
df['body_type'] = df['body_type'].fillna(
    df['body_type'].mode()[0]
)

df['transmission'] = df['transmission'].fillna(
    df['transmission'].mode()[0]
)
```

```python
df_same = df.dropna(subset=['original_price']).copy()
```

```python
print(df_same.shape)
print(df_same.isnull().sum())
```

```python
y = df_same['sale_price']
```

```python
x_without = df_same.drop(
    ['sale_price', 'original_price'],
    axis=1
)
```

```python
x_with = df_same.drop(
    'sale_price',
    axis=1
)
```

```python
categorical = [
    'fuel_type',
    'city',
    'body_type',
    'transmission',
    'make',
    'model'
]

x_without = pd.get_dummies(
    x_without,
    columns=categorical,
    dtype=int
)

x_with = pd.get_dummies(
    x_with,
    columns=categorical,
    dtype=int
)
```

```python
train_idx, test_idx = train_test_split(
    df_same.index,
    test_size=0.2,
    random_state=42
)
```

```python
x_without_train = x_without.loc[train_idx]
x_without_test = x_without.loc[test_idx]

x_with_train = x_with.loc[train_idx]
x_with_test = x_with.loc[test_idx]

y_train = y.loc[train_idx]
y_test = y.loc[test_idx]
```

```python
lr_without = LinearRegression()

lr_without.fit(
    x_without_train,
    y_train
)

pred_without = lr_without.predict(
    x_without_test
)
```

```python
print("WITHOUT ORIGINAL PRICE")
print("MAE:", metrics.mean_absolute_error(y_test, pred_without))
print("RMSE:", metrics.root_mean_squared_error(y_test, pred_without))
print("R2:", metrics.r2_score(y_test, pred_without))
```

```python
lr_with = LinearRegression()

lr_with.fit(
    x_with_train,
    y_train
)

pred_with = lr_with.predict(
    x_with_test
)
```

```python
print("Advanced model")
print("WITH ORIGINAL PRICE")
print("MAE:", metrics.mean_absolute_error(y_test, pred_with))
print("RMSE:", metrics.root_mean_squared_error(y_test, pred_with))
print("R2:", metrics.r2_score(y_test, pred_with))
```
